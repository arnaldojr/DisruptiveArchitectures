## 5. Funções

Uma função reúne uma responsabilidade sob um nome. Ela pode receber **parâmetros** e devolver um valor com `return`.

### `print()` não é `return`

- `print()` mostra algo na tela.
- `return` entrega um valor para quem chamou a função.

```python
def mostrar_titulo(titulo: str):
    print("Título:", titulo)


def obter_titulo(livro: dict) -> str:
    return livro["titulo"]


retorno_do_print = mostrar_titulo("Sapiens")
retorno_da_funcao = obter_titulo(CATALOGO[0])

print("Retorno de mostrar_titulo:", retorno_do_print)
print("Retorno de obter_titulo:", retorno_da_funcao)
```

### Escopo local

Uma variável criada dentro da função pertence à execução daquela função. Ela pode usar o mesmo nome de uma variável externa sem substituir automaticamente o valor externo.

```python
mensagem = "fora da função"


def criar_mensagem(titulo: str) -> str:
    mensagem = "Livro selecionado: " + titulo
    return mensagem


resultado = criar_mensagem("Eu, Robô")
print(resultado)
print(mensagem)
```

### Função de consulta

Agora transformamos a busca anterior em uma função reutilizável.

```python
def consultar_livro(id_livro: int) -> dict:
    for item in CATALOGO:
        if item["id"] == id_livro:
            return item

    return {"erro": "livro não encontrado", "id_livro": id_livro}


print(consultar_livro(1))
print(consultar_livro(999))
```

4. crie um identificador;
5. adicione o livro ao catálogo;

```python
# Escreva sua tentativa aqui.
#
# def cadastrar_livro(titulo, autor, categoria):
#     ...
```

### Solução possível

A implementação abaixo usa passos explícitos. Leia cada validação antes de executar.

```python
def cadastrar_livro(titulo: str, autor: str, categoria: str) -> dict:
    titulo_limpo = titulo.strip()
    autor_limpo = autor.strip()

    if len(titulo_limpo) < 2:
        return {"erro": "título inválido"}

    if len(autor_limpo) < 2:
        return {"erro": "autor inválido"}

    if categoria not in categorias:
        return {"erro": "categoria inválida"}

    maior_id = 0
    for item in CATALOGO:
        if item["id"] > maior_id:
            maior_id = item["id"]

    novo_livro = {
        "id": maior_id + 1,
        "titulo": titulo_limpo,
        "autor": autor_limpo,
        "categoria": categoria,
        "disponivel": True,
    }

    CATALOGO.append(novo_livro)
    return novo_livro
```

### Teste entradas válidas e inválidas

Observe que uma entrada inválida devolve um erro e não altera o catálogo.

```python
quantidade_antes = len(CATALOGO)

cadastro_valido = cadastrar_livro("Duna", "Frank Herbert", "ficcao")
cadastro_sem_titulo = cadastrar_livro(" ", "Autor Teste", "ficcao")
cadastro_categoria_invalida = cadastrar_livro(
    "Livro Teste", "Autor Teste", "aventura"
)

print("Cadastro válido:", cadastro_valido)
print("Título inválido:", cadastro_sem_titulo)
print("Categoria inválida:", cadastro_categoria_invalida)
print("Livros adicionados:", len(CATALOGO) - quantidade_antes)
```

### Liste apenas os disponíveis

Criamos uma nova lista e adicionamos somente os itens que atendem à condição.

```python
def listar_disponiveis() -> list[dict]:
    disponiveis = []

    for item in CATALOGO:
        if item["disponivel"]:
            disponiveis.append(item)

    return disponiveis


for item in listar_disponiveis():
    print(item["titulo"])
```

### Checks das funções

`assert` verifica uma condição. Se ela for falsa, o Python interrompe a execução com `AssertionError`.

```python
assert consultar_livro(1)["titulo"] == "Eu, Robô"
assert consultar_livro(999)["erro"] == "livro não encontrado"
assert cadastro_valido["titulo"] == "Duna"
assert cadastro_categoria_invalida["erro"] == "categoria inválida"

for item in listar_disponiveis():
    assert item["disponivel"] is True

print("Checks das funções: OK")
```

## 6. Orientação a objetos essencial

Uma **classe** descreve quais dados e comportamentos seus objetos terão. Cada objeto criado a partir dela é uma **instância**.

`__init__` recebe os valores iniciais. `self` representa o próprio objeto.

```python
class Livro:
    def __init__(self, id_livro: int, titulo: str, autor: str, categoria: str, disponivel: bool = True):
        self.id_livro = id_livro
        self.titulo = titulo
        self.autor = autor
        self.categoria = categoria
        self.disponivel = disponivel

    def resumo(self) -> str:
        if self.disponivel:
            status = "disponível"
        else:
            status = "indisponível"

        return f"{self.titulo}, de {self.autor} — {status}"
```

### Crie objetos

Cada chamada a `Livro(...)` cria um objeto diferente, com seus próprios atributos.

```python
livro_objeto_1 = Livro(10, "Neuromancer", "William Gibson", "ficcao")
livro_objeto_2 = Livro(11, "Código Limpo", "Robert C. Martin", "tecnologia", False )

print(livro_objeto_1.titulo)
print(livro_objeto_1.resumo())
print(livro_objeto_2.resumo())
```

### Defina um modelo

`LivroValidado` herda recursos de `BaseModel`. Os type hints definem os tipos esperados e `Field` acrescenta restrições.

```python
from typing import Literal
from pydantic import BaseModel, Field, ValidationError


class LivroValidado(BaseModel):
    id_livro: int = Field(gt=0)
    titulo: str = Field(min_length=2)
    autor: str = Field(min_length=2)
    categoria: Literal["ficcao", "tecnologia", "historia", "outro"]
    disponivel: bool = True
```

### Validação bem-sucedida

Quando os dados atendem às regras, recebemos um objeto. Seus campos podem ser acessados como atributos.

```python
livro_validado = LivroValidado(
    id_livro=20,
    titulo="Fundação",
    autor="Isaac Asimov",
    categoria="ficcao",
    disponivel=False
)

print(livro_validado)
print("Título:", livro_validado.titulo)
print("Disponível:", livro_validado.disponivel)
```

### `model_dump()`

O objeto validado pode voltar a ser um dicionário. Esse método aparecerá novamente em aplicações que precisam enviar dados para outros componentes.

```python
livro_como_dicionario = livro_validado.model_dump()

print(livro_como_dicionario)
print(type(livro_como_dicionario))
```

### Trate um `ValidationError`

Dados inválidos geram uma exceção. `try/except` permite que a aplicação trate esse problema em vez de encerrar inesperadamente.

```python
try:
    livro_invalido = LivroValidado(
        id_livro=0,
        titulo="X",
        autor="",
        categoria="aventura",
    )
    print(livro_invalido)
except ValidationError as erro:
    print("Os dados não passaram pela validação.")
    print(erro)
```

### Inspecione os erros

`erro.errors()` devolve uma lista de dicionários. Assim, a aplicação pode identificar o campo e a mensagem de cada erro.

```python
try:
    LivroValidado(
        id_livro=-5,
        titulo="A",
        autor="B",
        categoria="desconhecida",
    )
except ValidationError as erro:
    erros = erro.errors()

    for item_erro in erros:
        print("Campo:", item_erro["loc"])
        print("Mensagem:", item_erro["msg"])
        print()
```

### Validação manual e Pydantic

Na função `cadastrar_livro`, escrevemos cada regra com `if`. No modelo Pydantic, os campos e restrições formam um contrato reutilizável.

Pydantic reduz parte do código repetitivo, mas a aplicação ainda precisa definir regras de negócio, como verificar duplicidade ou disponibilidade.

## 8. Python e JSON

JSON é um formato textual usado para transportar dados. O módulo `json` faz parte da biblioteca padrão do Python.

```python
import json


dados_python = {
    "titulo": "Fundação",
    "disponivel": True,
    "data_devolucao": None,
    "tags": ["ficcao", "classico"],
}

print(dados_python)
print(type(dados_python))
```

### Python → JSON com `dumps()`

O resultado é uma string. Observe `true` e `null`, que pertencem à sintaxe JSON.

### Check da ida e volta

Depois da conversão Python → JSON → Python, os dados devem continuar equivalentes.

```python
assert dados_recuperados == dados_python
print("Conversão Python → JSON → Python: OK")
```

## 9. Miniaplicação: execute uma solicitação de ferramenta

Primeiro, definimos modelos Pydantic específicos para os argumentos de cada operação.

```python
class ArgumentosConsulta(BaseModel):
    id_livro: int = Field(gt=0)


class ArgumentosCadastro(BaseModel):
    titulo: str = Field(min_length=2)
    autor: str = Field(min_length=2)
    categoria: Literal["ficcao", "tecnologia", "historia", "outro"]
```

### Controle quais funções podem ser executadas

A função abaixo usa `if/elif`. Somente três nomes são reconhecidos. Qualquer outro nome será recusado.

As chamadas usam argumentos explícitos para deixar o fluxo visível.

- **model_validate**: é um metodo do Pydantic que valida os argumentos recebidos. Ele devolve um objeto com os campos validados ou gera uma exceção.

```python
def executar_solicitacao(solicitacao: dict) -> dict:
    nome = solicitacao.get("nome")
    argumentos = solicitacao.get("argumentos", {})

    if nome == "consultar_livro":
        try:
            dados = ArgumentosConsulta.model_validate(argumentos)
        except ValidationError as erro:
            return {
                "sucesso": False,
                "erro": "argumentos inválidos",
                "detalhes": str(erro),
            }

        resultado = consultar_livro(dados.id_livro)

        if "erro" in resultado:
            return {"sucesso": False, "resultado": resultado}

        return {"sucesso": True, "resultado": resultado}

    elif nome == "cadastrar_livro":
        try:
            dados = ArgumentosCadastro.model_validate(argumentos)
        except ValidationError as erro:
            return {
                "sucesso": False,
                "erro": "argumentos inválidos",
                "detalhes": str(erro),
            }

        resultado = cadastrar_livro(
            dados.titulo,
            dados.autor,
            dados.categoria,
        )

        if "erro" in resultado:
            return {"sucesso": False, "resultado": resultado}

        return {"sucesso": True, "resultado": resultado}

    elif nome == "listar_disponiveis":
        resultado = listar_disponiveis()
        return {"sucesso": True, "resultado": resultado}

    else:
        return {
            "sucesso": False,
            "erro": "ferramenta não autorizada",
            "nome": nome,
        }
```

### Teste uma ferramenta conhecida

```python
solicitacao_consulta = {
    "nome": "consultar_livro",
    "argumentos": {"id_livro": 2},
}

resultado_consulta = executar_solicitacao(solicitacao_consulta)
print(resultado_consulta)
```

```python
solicitacao_invalida = {
    "nome": "consultar_livro",
    "argumentos": {"id_livro": 0},
}

resultado_invalido = executar_solicitacao(solicitacao_invalida)
print(resultado_invalido)
```
